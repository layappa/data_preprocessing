```python
import pandas as pd
import  matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import seaborn as sns 
```

```python
dataset=pd.read_csv(r'R:\aiml\ml\loan_dataset.csv')
```

```python
dataset.describe()
```

```python
ss=StandardScaler()
dataset["scaled_ApplicantIncome"]=ss.fit_transform(dataset[['ApplicantIncome']])
```

```python
dataset.describe()
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("before")
sns.distplot(dataset['ApplicantIncome'])
plt.subplot(1,2,2)
plt.title("after")
sns.distplot(dataset['scaled_ApplicantIncome'])
plt.show()
```
